```python
from IPython import get_ipython
from IPython.display import display
# %%
import cv2
import numpy as np
# Import the necessary function for displaying images in Colab
from google.colab.patches import cv2_imshow

# Load an image
image = cv2.imread('/content/ganeshji.jpeg', cv2.IMREAD_GRAYSCALE)

# Apply Gaussian Blur to reduce noise and improve edge detection
blurred_image = cv2.GaussianBlur(image, (5, 5), 0)

# Apply Canny edge detection
edges = cv2.Canny(blurred_image, threshold1=100, threshold2=200)

# Display the results using cv2_imshow instead of cv2.imshow
cv2_imshow(image)  # Display original image
cv2_imshow(edges)  # Display edges

cv2.waitKey(0)
cv2.destroyAllWindows()
```
